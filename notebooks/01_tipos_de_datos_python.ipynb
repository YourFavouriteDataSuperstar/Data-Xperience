## 1. Introducción a los tipos de datos

En Python, cada valor tiene un **tipo de dato** asociado, que determina qué operaciones se pueden hacer con él y cómo se comporta. Por ejemplo, no tiene sentido "sumar" el texto `"hola"` con el número `5` sin antes convertir alguno de los dos: son tipos distintos y Python no sabe cómo combinarlos automáticamente.

Python es un lenguaje de **tipado dinámico**: no es necesario declarar el tipo de una variable antes de usarla (a diferencia de lenguajes como Java o C++). El intérprete infiere el tipo automáticamente a partir del valor que le asignas. Esto hace que Python sea muy flexible para aprender y prototipar rápido, pero también implica que debemos estar atentos a qué tipo tiene cada variable en cada momento, porque puede cambiar.

Usamos la función `type()` para saber qué tipo de dato tiene una variable en un momento dado. Esta función es una de las herramientas de depuración más simples y más usadas: cuando algo no funciona como esperas, preguntarte "¿de qué tipo es esta variable?" suele ser el primer paso para entender el error.


````python
# Ejemplo: inspeccionar el tipo de distintas variables
edad = 21
estatura = 1.68
nombre = "Camila"
es_estudiante = True

print(type(edad))
print(type(estatura))
print(type(nombre))
print(type(es_estudiante))
````

Fíjate en la salida: Python te devuelve algo como `<class 'int'>`. La palabra "class" (clase) aparece porque, internamente, todos los tipos en Python son en realidad clases, y cada variable es una instancia de esa clase. No necesitas profundizar en programación orientada a objetos para usar esto: por ahora basta con leer `<class 'int'>` como "esta variable es un entero".

> **Observación clave — el tipado dinámico:** una misma variable puede cambiar de tipo si le asignamos un valor de otro tipo. Esto es distinto de lenguajes con tipado estático, donde una variable declarada como entero nunca podría convertirse en texto.


````python
x = 10          # x es int
print(x, type(x))

x = "diez"      # ahora x es str
print(x, type(x))
````

Esto es poderoso, pero también puede ser una fuente de errores si no estás prestando atención: si en un programa largo reutilizas el nombre `x` para cosas distintas, puedes terminar operando sobre un tipo que no esperabas. Una buena práctica es usar nombres de variable descriptivos (`edad`, `precio_total`, `nombre_estudiante`) en lugar de nombres genéricos como `x`, precisamente para reducir esta confusión.

## 2. Tipos numéricos: `int` y `float`

Python tiene dos tipos numéricos principales que usarás constantemente:

- **`int`** (integer): números enteros, positivos o negativos, sin parte decimal. Ejemplos: `-5`, `0`, `42`, `1000000`.
- **`float`** (floating point): números con parte decimal, representados internamente en punto flotante. Ejemplos: `3.14`, `-0.5`, `2.0`.

También existen los **`complex`** (números complejos, con parte imaginaria), pero son poco usados en análisis de datos introductorio, así que no los cubriremos aquí.

> **Nota:** aunque `2` y `2.0` representan matemáticamente el mismo valor, en Python son de **tipos distintos** (`int` y `float` respectivamente), y eso puede afectar el resultado de ciertas operaciones o comparaciones en contextos más avanzados (por ejemplo, al trabajar con estructuras de datos que distinguen tipos, o al exportar datos a otros formatos).


````python
# Ejemplos de int y float
edad = 20                 # int
precio = 12500.50         # float
temperatura = -3.5        # float negativo
poblacion = 5_200_000      # los guiones bajos ayudan a leer números grandes (no afectan el valor)

print(edad, type(edad))
print(precio, type(precio))
print(temperatura, type(temperatura))
print(poblacion, type(poblacion))
````

### Operaciones aritméticas básicas

| Operador | Significado      | Ejemplo  | Resultado |
|:--------:|-------------------|:--------:|:---------:|
| `+`      | Suma               | `5 + 2`  | `7`       |
| `-`      | Resta              | `5 - 2`  | `3`       |
| `*`      | Multiplicación     | `5 * 2`  | `10`      |
| `/`      | División (float)   | `5 / 2`  | `2.5`     |
| `//`     | División entera    | `5 // 2` | `2`       |
| `%`      | Módulo (residuo)   | `5 % 2`  | `1`       |
| `**`     | Potencia           | `5 ** 2` | `25`      |

Vale la pena detenerse en las tres últimas filas, porque suelen confundir al principio:

- **`/` (división normal)** siempre devuelve un `float`, incluso si el resultado es un número exacto. Por ejemplo, `10 / 2` da `5.0`, no `5`. Esto es intencional: Python quiere que la división sea consistente y nunca pierda la parte decimal, aunque en ese caso particular sea cero.
- **`//` (división entera o "piso")** descarta la parte decimal del resultado y se queda solo con la parte entera hacia abajo. Es útil, por ejemplo, para calcular cuántos grupos completos de cierto tamaño caben en una cantidad (si tienes 17 estudiantes y quieres grupos de 5, `17 // 5` te dice que caben 3 grupos completos).
- **`%` (módulo)** te da el residuo de una división entera. Es muy útil para saber si un número es par o impar (`n % 2 == 0` es par), o para trabajar con ciclos (por ejemplo, determinar en qué día de la semana cae algo).


````python
a = 17
b = 5

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)
print("División entera:", a // b)
print("Módulo:", a % b)
print("Potencia:", a ** b)
````

## 3. Cadenas de texto: `str`

Las cadenas (`str`, de *string*) representan texto: nombres, direcciones, mensajes, códigos, etc. Se escriben entre comillas simples `'...'` o dobles `"..."` — en Python ambas son equivalentes, la elección es una cuestión de estilo (aunque es útil usar comillas dobles si el texto contiene un apóstrofe, y viceversa).

> **Nota:** los strings son **inmutables**. Esto significa que, una vez creado un string, no puedes cambiar un carácter individual dentro de él (por ejemplo, no puedes hacer `nombre[0] = "J"` para cambiar la primera letra). En cambio, los métodos de string como `.upper()` o `.replace()` no modifican el string original: **crean y devuelven uno nuevo**. Por eso siempre debes guardar el resultado en una variable (la misma u otra) si quieres conservarlo.


````python
curso = "DataXperience"
universidad = 'EAN'
frase = f"Estudio {curso} en la {universidad}"  # f-string: permite insertar variables dentro del texto

print(curso, type(curso))
print(frase)
````

Las **f-strings** (cadenas con formato, escritas como `f"..."`) son muy útiles porque te permiten insertar el valor de una variable directamente dentro de un texto, poniendo el nombre de la variable entre llaves `{}`. Son la forma moderna y recomendada de construir mensajes que combinan texto fijo con valores que cambian.

### Operaciones y métodos comunes con `str`

- **Concatenación (`+`)**: une dos strings en uno solo.
- **Repetición (`*`)**: repite un string un número de veces.
- **Longitud (`len()`)**: cuenta cuántos caracteres tiene el string (incluyendo espacios).
- **Métodos útiles**: `.upper()` (mayúsculas), `.lower()` (minúsculas), `.strip()` (elimina espacios en blanco al inicio y al final), `.replace(viejo, nuevo)` (reemplaza texto), `.split(separador)` (divide el string en una lista de partes).


````python
saludo = "hola"
nombre = "Mariana"

print(saludo + ", " + nombre)      # concatenación
print(saludo * 3)                  # repetición
print(len(nombre))                 # longitud
print(nombre.upper())              # mayúsculas
print(nombre.lower())              # minúsculas
print("  espacios de sobra  ".strip())  # quitar espacios
print(nombre.replace("a", "@"))    # reemplazar caracteres
print("Python,Pandas,NumPy".split(","))  # dividir en lista
````

## 4. Booleanos: `bool`

El tipo `bool` solo tiene dos valores posibles: `True` o `False` (con mayúscula inicial, es sensible a mayúsculas/minúsculas). Se usan para representar condiciones lógicas, y son el resultado natural de cualquier comparación.

Los booleanos son fundamentales porque son la base de las estructuras de control que verás más adelante en el curso (`if`, `while`): cualquier condición que evalúes ("¿el promedio es mayor a 3.0?", "¿el estudiante está activo?") se resuelve internamente en un valor `bool`.


````python
mayor_de_edad = 20 >= 18
tiene_beca = False

print(mayor_de_edad, type(mayor_de_edad))
print(tiene_beca, type(tiene_beca))

# Operadores de comparación
print(5 == 5)   # igualdad (dos signos "="; uno solo se usa para asignar, no para comparar)
print(5 != 3)   # distinto
print(5 > 3)
print(5 < 3)

# Operadores lógicos: combinan condiciones booleanas
print(True and False)  # "and": verdadero solo si AMBOS lados son verdaderos
print(True or False)   # "or": verdadero si AL MENOS uno de los lados es verdadero
print(not True)        # "not": invierte el valor
````

> **Cuidado:** un error muy común de principiantes es confundir `=` (asignación, "guarda este valor en esta variable") con `==` (comparación, "¿son iguales estos dos valores?"). Si usas `=` donde querías comparar, Python en muchos contextos lanzará un error de sintaxis, pero es importante tener clara la diferencia conceptual.

> **Dato curioso:** en Python, `bool` es en realidad una subclase de `int`. Por eso `True` se comporta como `1` y `False` como `0` en operaciones aritméticas. Esto no es solo una curiosidad: en la práctica te permite, por ejemplo, sumar una lista de booleanos para contar cuántos son `True` (una técnica muy usada al analizar datos).


````python
print(True + True)     # 2
print(False * 10)      # 0
print(isinstance(True, int))  # True: confirma que bool es un tipo de int
````

## 5. Estructuras de datos: `list`, `tuple`, `dict`, `set`

Además de los tipos simples que representan un único valor, Python ofrece **estructuras de datos** para agrupar varios valores relacionados en una sola variable. Elegir la estructura correcta depende de tres preguntas: ¿necesito que el orden de los elementos se mantenga?, ¿necesito poder modificar la colección después de creada?, ¿pueden repetirse los elementos?

| Estructura | Sintaxis    | ¿Ordenada?         | ¿Modificable? | ¿Duplicados?  | Cuándo usarla |
|:----------:|:-----------:|:------------------:|:--------------:|:--------------:|---------------|
| `list`     | `[1, 2, 3]` | Sí                  | Sí              | Sí              | Colección de elementos que puede crecer, encogerse o reordenarse (por ejemplo, una lista de calificaciones). |
| `tuple`    | `(1, 2, 3)` | Sí                  | No              | Sí              | Datos agrupados que no deberían cambiar (por ejemplo, unas coordenadas geográficas fijas). |
| `dict`     | `{"a": 1}`  | Sí (desde Python 3.7) | Sí            | Claves únicas   | Asociar una clave descriptiva con un valor (por ejemplo, los datos de un estudiante identificados por campo). |
| `set`      | `{1, 2, 3}` | No                  | Sí              | No              | Verificar pertenencia o eliminar duplicados rápidamente, sin importar el orden. |

Piensa en la diferencia entre `list` y `tuple` como la diferencia entre una lista de tareas (que vas tachando y agregando) y las coordenadas de tu casa (que no cambian, así que tiene sentido protegerlas de modificaciones accidentales).


````python
# Listas: colecciones ordenadas y modificables
notas = [4.5, 3.8, 5.0, 2.9]
print(notas, type(notas))

notas.append(4.0)     # agregar un elemento al final
print(notas)

print(notas[0])        # primer elemento (Python empieza a contar desde el índice 0)
print(notas[-1])       # último elemento (los índices negativos cuentan desde el final)
````

````python
# Tuplas: como listas, pero inmutables
coordenadas = (4.6486, -74.0908)  # ejemplo: latitud, longitud
print(coordenadas, type(coordenadas))

# coordenadas[0] = 10   # si descomentas esta línea, obtendrás un TypeError:
                         # las tuplas no permiten asignar un nuevo valor a una posición
````

````python
# Diccionarios: pares clave-valor
estudiante = {
    "nombre": "Laura",
    "edad": 22,
    "programa": "Administración de Empresas"
}
print(estudiante, type(estudiante))

print(estudiante["nombre"])   # acceder al valor a través de su clave
estudiante["semestre"] = 5     # agregar una nueva clave-valor
print(estudiante)
````

````python
# Conjuntos: colecciones sin orden ni duplicados
lenguajes = {"Python", "R", "SQL", "Python"}  # "Python" está duplicado a propósito
print(lenguajes, type(lenguajes))  # el duplicado desaparece automáticamente

otro_grupo = {"SQL", "Excel"}
print(lenguajes.union(otro_grupo))         # unión: todos los elementos de ambos conjuntos
print(lenguajes.intersection(otro_grupo))  # intersección: solo los elementos que están en ambos
````

## 6. Conversión de tipos (*type casting*)

A veces necesitamos convertir un dato de un tipo a otro. Para eso usamos funciones como `int()`, `float()`, `str()`, `bool()`, `list()`, entre otras. A este proceso se le llama *type casting* o conversión de tipos.

Esto es extremadamente común cuando trabajas con datos reales: por ejemplo, al leer un archivo CSV o al recibir información de un formulario web, **todo llega inicialmente como texto (`str`)**, incluso los números. Si quieres sumar, promediar o comparar esos valores numéricamente, primero debes convertirlos explícitamente al tipo correcto (`int` o `float`). Este es uno de los pasos de limpieza de datos más frecuentes en cualquier proyecto de análisis de datos.


````python
edad_texto = "25"
edad_numero = int(edad_texto)
print(edad_numero, type(edad_numero))
print(edad_numero + 5)   # ahora sí podemos sumar, porque es un número y no texto

precio_texto = "19999.99"
precio_numero = float(precio_texto)
print(precio_numero, type(precio_numero))

numero = 42
numero_texto = str(numero)
print(numero_texto, type(numero_texto))
````

> **Cuidado:** no todas las conversiones son posibles. Si el texto no representa un número válido (por ejemplo, contiene letras o símbolos que no son parte de un número), Python lanza un error de tipo `ValueError` y detiene la ejecución en ese punto. Por eso, en código que procesa datos externos (que pueden venir "sucios" o mal formateados), es común usar bloques `try/except` para capturar y manejar este tipo de errores sin que el programa se detenga abruptamente.


````python
try:
    int("veinte")
except ValueError as error:
    print("Error al convertir:", error)
````

## Resumen

- Python tiene tipos de datos simples: `int`, `float`, `str`, `bool`, cada uno con reglas propias de comportamiento (por ejemplo, los strings son inmutables, y `/` siempre da `float`).
- También tiene estructuras para agrupar datos: `list` (ordenada y modificable), `tuple` (ordenada e inmutable), `dict` (pares clave-valor) y `set` (sin orden ni duplicados). Elegir la estructura correcta depende de si necesitas orden, modificabilidad o unicidad.
- La función `type()` te permite inspeccionar el tipo de cualquier variable en cualquier momento, y es una herramienta de depuración esencial.
- La conversión de tipos (*type casting*, con `int()`, `float()`, `str()`, `bool()`) es indispensable al trabajar con datos que llegan como texto, como ocurre casi siempre al leer archivos o formularios. Presta especial atención a la conversión a `bool`, que puede dar resultados inesperados si conviertes directamente desde un string.
