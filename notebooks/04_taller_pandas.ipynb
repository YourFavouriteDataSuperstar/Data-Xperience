```python
# ============================================================
#  DATOS DEL TALLER - EJECUTA ESTA CELDA PRIMERO
#  DataFood: operacion de domicilios, marzo a junio de 2024
#  No modifiques esta celda.
# ============================================================
import pandas as pd
import numpy as np

np.random.seed(11)
n = 520

ciudades   = ["Bogota", "Medellin", "Cali", "Barranquilla", "Bucaramanga"]
categorias = ["Comida rapida", "Saludable", "Asiatica", "Postres", "Tipica", "Cafe"]
canales    = ["App", "Web", "Telefono"]
pagos      = ["Tarjeta", "Efectivo", "Billetera digital"]

ids_rest = [f"R-{i:03d}" for i in range(1, 41)]
ids_rep  = [f"D-{i:02d}" for i in range(1, 31)]

dias_base = pd.date_range("2024-03-01", periods=120, freq="D")
peso_dia  = np.select([dias_base.dayofweek == 4, dias_base.dayofweek == 5, dias_base.dayofweek == 6],
                      [1.4, 1.9, 1.5], default=1.0)
peso_dia  = peso_dia / peso_dia.sum()
peso_hora = np.array([.4, .2, .15, .1, .1, .25, .7, 1.1, 1.4, 1.3, 1.6, 3.2,
                      6.8, 7.2, 4.4, 2.4, 2.1, 3.1, 5.6, 7.6, 6.1, 3.4, 1.9, .9])
peso_hora = peso_hora / peso_hora.sum()

fecha_hora = (pd.to_datetime("2024-03-01")
              + pd.to_timedelta(np.random.choice(120, n, p=peso_dia), unit="D")
              + pd.to_timedelta(np.random.choice(24, n, p=peso_hora) * 60 + np.random.randint(0, 60, n), unit="m"))

ciudad = np.random.choice(ciudades, n, p=[.34, .24, .18, .14, .10])
factor_ciudad = pd.Series({"Bogota": 1.18, "Medellin": 1.06, "Cali": 0.94,
                           "Barranquilla": 0.88, "Bucaramanga": 0.78})

valor  = np.round(np.random.gamma(3.2, 14000, n) * factor_ciudad[ciudad].values + 8000, -2)
estado = np.random.choice(["Entregado", "Cancelado"], n, p=[0.92, 0.08])

minutos = np.round(np.random.normal(38, 11, n)).clip(12, 95)
minutos[np.random.choice(np.arange(n), 7, replace=False)] = np.random.choice([180, 210, 240], 7)

pedidos = pd.DataFrame({
    "id_pedido":       range(90001, 90001 + n),
    "fecha_hora":      fecha_hora,
    "ciudad":          ciudad,
    "id_restaurante":  np.random.choice(ids_rest, n),
    "categoria":       np.random.choice(categorias, n, p=[.30, .15, .18, .12, .15, .10]),
    "canal":           np.random.choice(canales, n, p=[.62, .28, .10]),
    "metodo_pago":     np.random.choice(pagos, n, p=[.45, .25, .30]),
    "valor_productos": valor,
    "costo_envio":     np.random.choice([0, 2900, 4500, 6900, 9900], n, p=[.15, .25, .30, .20, .10]),
    "propina":         np.where(np.random.rand(n) < 0.42,
                                np.round(valor * np.random.choice([.05, .08, .10], n), -2), 0.0),
    "minutos_entrega": minutos,
    "id_repartidor":   np.random.choice(ids_rep[:26], n),
    "calificacion":    np.random.choice([1., 2., 3., 4., 5.], n, p=[.04, .07, .19, .38, .32]),
    "estado":          estado,
})

# --- Asi salio del sistema. Es parte del ejercicio. ---
cancel = pedidos["estado"] == "Cancelado"
pedidos.loc[cancel, ["minutos_entrega", "id_repartidor", "calificacion", "propina"]] = np.nan
pedidos.loc[np.random.choice(pedidos.index[~cancel], 24, replace=False), "calificacion"] = np.nan

pedidos["minutos_entrega"] = pedidos["minutos_entrega"].astype("object")
pedidos.loc[pedidos["minutos_entrega"].isna(), "minutos_entrega"] = ""
pedidos["minutos_entrega"] = pedidos["minutos_entrega"].apply(lambda x: "" if x == "" else str(int(x)))
pedidos.loc[np.random.choice(pedidos.index[~cancel], 6, replace=False), "minutos_entrega"] = "N/D"
sucios_min = np.random.choice(pedidos.index[~cancel], 10, replace=False)
pedidos.loc[sucios_min, "minutos_entrega"] = "  " + pedidos.loc[sucios_min, "minutos_entrega"].astype(str) + " "

sucios = np.random.choice(pedidos.index, 17, replace=False)
pedidos.loc[sucios, "ciudad"] = "  " + pedidos.loc[sucios, "ciudad"].str.lower() + "  "

pedidos = pd.concat([pedidos, pedidos.sample(14, random_state=3)], ignore_index=True)

restaurantes = pd.DataFrame({
    "id_restaurante":   ids_rest[:36],
    "nombre":           [f"Restaurante {i:02d}" for i in range(1, 37)],
    "tipo_cocina":      np.random.choice(categorias, 36),
    "comision_pct":     np.random.choice([0.12, 0.15, 0.18, 0.22], 36),
    "estrellas":        np.round(np.random.uniform(3.0, 5.0, 36), 1),
    "fecha_afiliacion": pd.to_datetime("2021-01-01") + pd.to_timedelta(np.random.randint(0, 1100, 36), unit="D"),
})

repartidores = pd.DataFrame({
    "id_repartidor":     ids_rep,
    "nombre_repartidor": [f"Repartidor {i:02d}" for i in range(1, 31)],
    "vehiculo":          np.random.choice(["Moto", "Bicicleta", "Carro"], 30, p=[.6, .3, .1]),
    "zona_asignada":     np.random.choice(["Norte", "Centro", "Sur", "Occidente"], 30),
    "fecha_ingreso":     pd.to_datetime("2022-06-01") + pd.to_timedelta(np.random.randint(0, 700, 30), unit="D"),
})

# Metas de ingreso de plataforma fijadas por gerencia para cada ciudad y mes
metas = pd.DataFrame({
    "ciudad": (["Bogota"] * 4 + ["Medellin"] * 4 + ["Cali"] * 4 +
               ["Barranquilla"] * 4 + ["Bucaramanga"] * 4 + ["Cartagena"] * 4),
    "mes": [3, 4, 5, 6] * 6,
    "meta_ingreso": [500000, 550000, 600000, 650000,
                     300000, 320000, 340000, 320000,
                     200000, 210000, 220000, 230000,
                     150000, 160000, 170000, 180000,
                      90000,  95000, 100000, 105000,
                     120000, 130000, 140000, 150000],
})

print("Datos cargados.")
print("pedidos:", pedidos.shape, "| restaurantes:", restaurantes.shape,
      "| repartidores:", repartidores.shape, "| metas:", metas.shape)
pedidos.head()
```
