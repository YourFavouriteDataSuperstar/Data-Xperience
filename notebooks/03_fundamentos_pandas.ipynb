**pandas** es la libreria estandar de Python para trabajar con **datos tabulares**: informacion organizada en filas y columnas, como la que ves en una hoja de Excel, en una tabla de base de datos o en un archivo CSV. Su nombre viene de *panel data*, un termino de econometria que designa datos observados para varias entidades a lo largo del tiempo.

Para entender por que existe, piensa en como resolverias esto solo con Python puro: tienes 50.000 registros de ventas y necesitas el promedio por ciudad. Tendrias que recorrer una lista de diccionarios con un `for`, ir acumulando sumas y conteos en otro diccionario, dividir al final y ordenar el resultado. Son unas 15 lineas, propensas a errores y lentas. En pandas es **una sola linea**, y por debajo se ejecuta en codigo C optimizado.

Esa es la promesa de pandas: **operaciones sobre tablas completas en lugar de recorridos elemento por elemento**. A esta forma de trabajar se le llama *operacion vectorizada*, y es la diferencia entre pensar "para cada fila, haz esto" y pensar "a esta columna entera, aplicale esto".

La comparacion mas util es con las estructuras que ya conoces de Python:

Una Series es, entonces, lo mejor de una lista y de un diccionario, mas la capacidad de hacer aritmetica sobre todos sus elementos a la vez. Ademas, a diferencia de una lista de Python, **todos los valores de una Series comparten el mismo tipo de dato** (`dtype`). Esa homogeneidad es la que permite que las operaciones sean rapidas.

```python
# --- Crear una Series a partir de una lista ---
ventas = pd.Series([1200, 850, 1430, 990, 1675])

print(ventas)
print()
print("Tipo del objeto :", type(ventas))
print("dtype de valores:", ventas.dtype)
```

Fijate en la salida. A la izquierda aparece una columna de numeros `0, 1, 2, 3, 4`: **ese es el indice**, que pandas creo automaticamente porque no le dimos uno. A la derecha estan los valores. Y abajo aparece `dtype: int64`, que indica que todos los valores son enteros de 64 bits.

El indice automatico funciona, pero pandas brilla cuando le das **etiquetas con significado**. Compara las dos formas: con indice numerico tienes que recordar que "la posicion 2 es Cali"; con indice de texto simplemente pides `ventas["Cali"]`.

```python
# --- Crear una Series con un indice significativo ---
ventas = pd.Series(
    [1200, 850, 1430, 990, 1675],                          # los valores
    index=["Bogota", "Medellin", "Cali", "Barranquilla", "Cartagena"],  # las etiquetas
    name="ventas_millones"                                  # nombre de la Series (sera el nombre de la columna)
)

print(ventas)
print()
print("Venta de Cali:", ventas["Cali"])          # acceso por ETIQUETA
print("Tercer valor :", ventas.iloc[2])          # acceso por POSICION (empieza en 0)
```

### Operaciones vectorizadas: el superpoder

Aqui esta la diferencia real con las listas de Python. Si tienes `lista = [1200, 850]` y escribes `lista * 1.19`, Python lanza un error. Si escribes `lista + lista`, Python **concatena** las listas en vez de sumar sus elementos.

Con una Series, en cambio, las operaciones aritmeticas se aplican **elemento por elemento sobre toda la estructura**, sin escribir un solo `for`. Esto se llama *vectorizacion*.

```python
# --- Operaciones sobre TODA la Series de una sola vez ---
print("Ventas originales:")
print(ventas)

print("\nVentas con IVA del 19% (multiplicacion elemento a elemento):")
print(ventas * 1.19)

print("\nDiferencia frente al promedio (resta de un escalar):")
print(ventas - ventas.mean())

print("\nParticipacion porcentual de cada ciudad:")
print(ventas / ventas.sum() * 100)
```

Observa que en ningun caso escribimos un ciclo. pandas aplica la operacion a los cinco valores simultaneamente y **conserva el indice**, de modo que el resultado sigue siendo interpretable: sabes que ese 21,5% corresponde a Cartagena.

### Metodos estadisticos incorporados

Toda Series numerica trae de fabrica un conjunto de metodos de resumen. Estos son los que mas vas a usar:

```python
# --- Estadisticos descriptivos de una Series ---
print("Suma total     :", ventas.sum())
print("Promedio       :", ventas.mean())
print("Mediana        :", ventas.median())
print("Desv. estandar :", ventas.std())
print("Minimo         :", ventas.min())
print("Maximo         :", ventas.max())
print("Cantidad       :", ventas.count())     # cuenta valores NO nulos

print("\nCiudad con MENOS ventas :", ventas.idxmin())   # devuelve la ETIQUETA, no el valor
print("Ciudad con MAS ventas   :", ventas.idxmax())

print("\nResumen completo con describe():")
print(ventas.describe())
```

### Filtrado booleano: la base de todo lo que sigue

Cuando comparas una Series con un valor, pandas no te devuelve `True` o `False`, sino **una Series completa de booleanos**, uno por cada elemento. Esa Series de booleanos se llama *mascara* y puede usarse como filtro entre corchetes.

Este mecanismo de dos pasos (crear la mascara, aplicar la mascara) es el fundamento de practicamente todo el filtrado que haras en pandas. Vale la pena que lo veas explicitamente separado antes de usarlo abreviado.

```python
# --- Paso 1: la comparacion produce una mascara booleana ---
mascara = ventas > 1000
print("La mascara (una Series de True/False):")
print(mascara)

# --- Paso 2: la mascara se usa como filtro ---
print("\nSolo las ciudades que cumplen la condicion:")
print(ventas[mascara])

# En la practica se escribe todo junto:
print("\nForma abreviada (identica, mas comun):")
print(ventas[ventas > 1000])
```

### El dataset de trabajo del notebook

Para que todo lo que sigue funcione sin depender de internet, vamos a **construir nuestro propio dataset**: las ventas de una cadena de tiendas de tecnologia durante 2024. Lo generamos con NumPy usando una semilla fija (`seed`), lo que garantiza que **todos obtengan exactamente los mismos numeros** y puedan comparar resultados.

Le introducimos a proposito algunos defectos tipicos de datos reales (valores nulos, filas duplicadas, una columna numerica guardada como texto), porque los vamos a necesitar en la seccion de limpieza. Un dataset perfecto no ensena nada.

```python
# --- Construccion del dataset de trabajo: ventas de una cadena de tecnologia ---
np.random.seed(42)   # semilla fija => resultados reproducibles para todo el grupo

n = 300   # numero de registros

ciudades  = ["Bogota", "Medellin", "Cali", "Barranquilla", "Cartagena", "Bucaramanga"]
regiones  = {"Bogota": "Andina", "Medellin": "Andina", "Cali": "Pacifica",
             "Barranquilla": "Caribe", "Cartagena": "Caribe", "Bucaramanga": "Andina"}
categorias = ["Computo", "Audio", "Accesorios", "Telefonia"]
vendedores = ["Ana", "Bruno", "Carla", "Diego", "Elena"]

ventas = pd.DataFrame({
    "id_venta":   range(1001, 1001 + n),
    "fecha":      pd.to_datetime("2024-01-01") + pd.to_timedelta(np.random.randint(0, 365, n), unit="D"),
    "ciudad":     np.random.choice(ciudades, n),
    "categoria":  np.random.choice(categorias, n, p=[0.35, 0.25, 0.25, 0.15]),
    "vendedor":   np.random.choice(vendedores, n),
    "unidades":   np.random.randint(1, 15, n),
    "precio_unitario": np.random.choice([85000, 190000, 450000, 1250000, 3200000], n),
    "descuento":  np.round(np.random.choice([0, 0, 0, 0.05, 0.10, 0.15], n), 2),
    "calificacion_cliente": np.random.choice([1, 2, 3, 4, 5, np.nan], n, p=[.05, .07, .18, .35, .30, .05]),
})

# La region se deduce de la ciudad usando un diccionario (veremos map() mas adelante)
ventas["region"] = ventas["ciudad"].map(regiones)

# --- Introducimos defectos a proposito, como en datos reales ---
ventas.loc[np.random.choice(ventas.index, 12, replace=False), "vendedor"] = np.nan   # nulos
ventas = pd.concat([ventas, ventas.sample(8, random_state=1)], ignore_index=True)    # duplicados
ventas["unidades"] = ventas["unidades"].astype(str)                                  # numero guardado como TEXTO

print("Dataset creado.")
print("Dimensiones:", ventas.shape)
ventas.head()
```

Cuando recibes un dataset por primera vez, no empiezas a calcular promedios: primero haces un **diagnostico**. El objetivo es responder cinco preguntas antes de tocar nada:

1. **Cuantos datos hay?** (`shape`)
2. **Que columnas tiene y de que tipo son?** (`info`)
3. **Como se ven los datos concretos?** (`head`, `tail`, `sample`)
4. **Como se distribuyen los numeros?** (`describe`)
5. **Que esta faltando o mal?** (`isna().sum()`, `duplicated().sum()`)

Saltarte este paso es la causa mas frecuente de analisis equivocados: promedios calculados sobre filas duplicadas, sumas que ignoran nulos, columnas numericas que en realidad son texto.

```python
# --- 1. Cuantos datos hay ---
print("Filas y columnas:", ventas.shape)
print("Total de celdas :", ventas.size)

# --- 2. Radiografia general: tipos y nulos por columna ---
print("\n" + "=" * 60)
ventas.info()
```

`info()` es el comando mas informativo de pandas y merece que aprendas a leerlo con detalle. Cada linea te dice tres cosas de una columna:

- **`Non-Null Count`**: cuantos valores no nulos tiene. Si este numero es menor que el total de filas, **esa columna tiene datos faltantes**. Compara: si el DataFrame tiene 308 filas y `vendedor` reporta 296 non-null, hay 12 nulos.
- **`Dtype`**: el tipo de dato. Aqui es donde detectas el problema clasico: `unidades` aparece como `object` cuando deberia ser numerica. Si intentas sumarla asi, pandas concatenara texto en vez de sumar.
- **`memory usage`** al final: cuanta RAM ocupa. Relevante cuando pasas de cientos de miles de filas.

```python
# --- 3. Ver los datos concretos ---
print("Primeras 5 filas (head):")
display(ventas.head())          # display() renderiza tablas con formato en Colab

print("Ultimas 3 filas (tail) - util para detectar filas de totales o basura al final:")
display(ventas.tail(3))

print("Muestra aleatoria de 5 filas (sample) - evita el sesgo de mirar siempre el principio:")
display(ventas.sample(5, random_state=7))
```

### `.loc` frente a `.iloc`: la distincion critica

- **`.loc[]`** trabaja con **etiquetas** (nombres de indice y de columna). Su rango **incluye el extremo final**.
- **`.iloc[]`** trabaja con **posiciones enteras** (0, 1, 2...). Su rango **excluye el extremo final**, igual que el slicing de listas de Python.

Esa diferencia en el extremo final es fuente constante de errores de un elemento. Recuerdalo asi: `.loc` habla el idioma de las etiquetas, donde pedir "de Bogota a Cali" incluye a Cali; `.iloc` habla el idioma de Python, donde `[0:3]` da tres elementos.

Ambos aceptan **dos argumentos separados por coma**: `df.loc[filas, columnas]`. Los dos puntos `:` significan "todo".

```python
# --- .loc: seleccion por ETIQUETA ---
print("Fila con etiqueta de indice 0:")
print(ventas.loc[0])

print("\n.loc con rango de filas 0 a 3 (INCLUYE el 3) y columnas nombradas:")
display(ventas.loc[0:3, ["ciudad", "categoria", "precio_unitario"]])

# --- .iloc: seleccion por POSICION ---
print(".iloc con rango 0:3 (EXCLUYE el 3 -> solo 3 filas) y columnas 2 a 4:")
display(ventas.iloc[0:3, 2:5])

print("Ultima fila del DataFrame (posicion -1, igual que en listas):")
print(ventas.iloc[-1])
```

### Filtrado por condiciones

Aqui aplicamos el mecanismo de mascaras booleanas que vimos en la seccion de Series, ahora sobre un DataFrame completo: la mascara se calcula sobre una columna, pero **filtra filas enteras**.

Cuando combinas varias condiciones hay dos reglas que **no puedes olvidar**:

1. Los operadores logicos son **`&` (y)**, **`|` (o)**, **`~` (no)**. **No** se usan `and`, `or`, `not` de Python, porque esos operan sobre un unico valor booleano y pandas necesita comparar elemento por elemento.
2. **Cada condicion debe ir entre parentesis**. Sin ellos, la precedencia de operadores de Python evalua `&` antes que `>` y obtienes un error confuso.

```python
# --- Filtro simple ---
caros = ventas[ventas["precio_unitario"] > 1000000]
print("Ventas con precio unitario > 1.000.000:", len(caros), "registros")

# --- Filtro con DOS condiciones: parentesis obligatorios, operador & ---
andina_computo = ventas[(ventas["region"] == "Andina") & (ventas["categoria"] == "Computo")]
print("Ventas de Computo en la region Andina:", len(andina_computo))

# --- Operador | (o) ---
costa = ventas[(ventas["ciudad"] == "Cartagena") | (ventas["ciudad"] == "Barranquilla")]
print("Ventas en Cartagena o Barranquilla:", len(costa))

# --- isin(): forma limpia de escribir muchos "o" encadenados ---
costa2 = ventas[ventas["ciudad"].isin(["Cartagena", "Barranquilla"])]
print("Lo mismo con isin():", len(costa2), "(identico, mas legible)")

# --- ~ para negar una condicion ---
no_costa = ventas[~ventas["ciudad"].isin(["Cartagena", "Barranquilla"])]
print("Ventas FUERA de la costa:", len(no_costa))

display(andina_computo.head(3))
```

### Filtros sobre texto y sobre nulos

Para columnas de texto, el accesor **`.str`** desbloquea todos los metodos de cadenas de Python aplicados a la columna entera: `.str.contains()`, `.str.startswith()`, `.str.upper()`, `.str.len()`. Para nulos, `.isna()` y `.notna()`.

```python
# --- Filtros de texto con el accesor .str ---
con_a = ventas[ventas["ciudad"].str.startswith("B")]
print("Ciudades que empiezan por B:", con_a["ciudad"].unique())

contiene = ventas[ventas["categoria"].str.contains("Tele", case=False)]
print("Categorias que contienen 'Tele':", contiene["categoria"].unique())

# --- Filtros de nulos ---
sin_vendedor = ventas[ventas["vendedor"].isna()]
print("\nRegistros SIN vendedor asignado:", len(sin_vendedor))

con_vendedor = ventas[ventas["vendedor"].notna()]
print("Registros CON vendedor asignado:", len(con_vendedor))
```

### `query()`: filtros que se leen como una frase

`query()` recibe la condicion como una cadena de texto y permite usar `and`, `or`, `not` en su forma natural. Para filtros con tres o mas condiciones resulta claramente mas legible. Para referirte a una variable de Python dentro de la consulta, se antepone `@`.

```python
# --- query(): mismo resultado, sintaxis mas legible ---
resultado_a = ventas[(ventas["region"] == "Andina") & (ventas["precio_unitario"] > 400000) & (ventas["descuento"] > 0)]
resultado_b = ventas.query("region == 'Andina' and precio_unitario > 400000 and descuento > 0")

print("Con corchetes:", len(resultado_a), "filas")
print("Con query()  :", len(resultado_b), "filas  <- identico")

# --- Usar una variable de Python dentro de query() con @ ---
umbral = 2000000
print("\nVentas por encima de", umbral, ":", len(ventas.query("precio_unitario > @umbral")))
```

### Problema 1: filas duplicadas

Un duplicado exacto casi siempre es un error de carga (el archivo se importo dos veces, un sistema envio el mismo registro repetido). Al eliminarlos, dos parametros importan:

- **`subset`**: por que columnas se define un duplicado. Sin `subset`, deben coincidir **todas** las columnas. Con `subset=["id_venta"]`, basta con que se repita el identificador.
- **`keep`**: cual conservar. `"first"` (por defecto) deja la primera aparicion, `"last"` la ultima, `False` elimina **todas** las copias incluida la original.

```python
# --- Eliminar duplicados ---
print("Antes :", len(df), "filas |", df.duplicated().sum(), "duplicados")

# Inspeccionar QUE se va a eliminar antes de eliminarlo (buena practica)
display(df[df.duplicated(keep=False)].sort_values("id_venta").head(6))

df = df.drop_duplicates()          # recuerda: hay que reasignar
print("\nDespues:", len(df), "filas |", df.duplicated().sum(), "duplicados")
```

### Problema 2: la columna numerica guardada como texto

`unidades` quedo como `object`. Esto ocurre constantemente en datos reales cuando un archivo trae simbolos de moneda, separadores de miles o un `"N/A"` perdido entre los numeros. Mientras siga siendo texto, `sum()` **concatenara cadenas** en lugar de sumar.

Hay dos herramientas para convertir:

- **`.astype(tipo)`**: conversion directa. Rapida, pero **falla con un error** si algun valor no se puede convertir.
- **`pd.to_numeric(serie, errors='coerce')`**: conversion tolerante. Los valores que no puede convertir los transforma en `NaN` en lugar de fallar. Es la opcion segura cuando no confias en la limpieza del origen.

```python
# --- Demostracion del problema ---
print("dtype actual:", df["unidades"].dtype)
print("Suma SIN convertir (concatena texto, no suma):", df["unidades"].sum()[:50], "...")

# --- Conversion segura con to_numeric ---
df["unidades"] = pd.to_numeric(df["unidades"], errors="coerce")

print("\ndtype despues :", df["unidades"].dtype)
print("Suma correcta  :", df["unidades"].sum())
print("Nulos generados por la conversion:", df["unidades"].isna().sum())
```

### Renombrar y estandarizar nombres de columnas

Los archivos reales llegan con nombres inconsistentes: mayusculas, espacios, tildes, caracteres especiales. Un nombre con espacio te impide usar la notacion `df.columna` y complica `query()`. Estandarizarlos al principio te ahorra fricciones durante todo el analisis.

```python
# --- Renombrar columnas especificas ---
df = df.rename(columns={"precio_unitario": "precio", "calificacion_cliente": "calificacion"})
print("Columnas ahora:", list(df.columns))

# --- Estandarizar TODOS los nombres de una vez (patron muy util) ---
ejemplo = pd.DataFrame({"Nombre Cliente": [1], "VALOR TOTAL ": [2], "Fecha-Compra": [3]})
print("\nAntes :", list(ejemplo.columns))

ejemplo.columns = (ejemplo.columns
                   .str.strip()            # quita espacios al inicio y al final
                   .str.lower()            # todo a minusculas
                   .str.replace(" ", "_")  # espacios -> guion bajo
                   .str.replace("-", "_")) # guiones -> guion bajo

print("Despues:", list(ejemplo.columns))
```

Los datos rara vez traen la variable que necesitas. Tienes precio y unidades, pero te preguntan por el ingreso total. Tienes fecha de nacimiento, pero necesitas rangos de edad. **Crear columnas derivadas es donde empieza el analisis de verdad.**

La forma mas directa es la asignacion: `df["nueva"] = expresion`. La expresion se evalua de forma vectorizada sobre todas las filas.

```python
# --- Columnas calculadas a partir de otras (operaciones vectorizadas) ---
df["ingreso_bruto"] = df["unidades"] * df["precio"]
df["valor_descuento"] = df["ingreso_bruto"] * df["descuento"]
df["ingreso_neto"] = df["ingreso_bruto"] - df["valor_descuento"]

display(df[["unidades", "precio", "descuento", "ingreso_bruto", "valor_descuento", "ingreso_neto"]].head())

print("Ingreso neto total de la cadena: $", f"{df['ingreso_neto'].sum():,.0f}")
```

### Condiciones: `np.where` para dos casos

Cuando la nueva columna depende de una condicion con **dos resultados posibles**, `np.where(condicion, valor_si_true, valor_si_false)` es la herramienta. Se lee como un `if/else` aplicado a toda la columna.

```python
# --- np.where: un if/else vectorizado ---
df["tuvo_descuento"] = np.where(df["descuento"] > 0, "Con descuento", "Sin descuento")

print(df["tuvo_descuento"].value_counts())

# Se pueden anidar para tres o mas casos, aunque pierde legibilidad rapidamente
df["tamano_venta"] = np.where(df["ingreso_neto"] > 5000000, "Grande",
                       np.where(df["ingreso_neto"] > 1000000, "Media", "Pequena"))

print("\n", df["tamano_venta"].value_counts())
```

### Rangos: `pd.cut` para convertir numeros en categorias

Cuando necesitas partir una variable numerica en intervalos (rangos de edad, rangos de precio, segmentos de valor), `pd.cut` es mas claro y menos propenso a errores que anidar `np.where`. Le das los **puntos de corte** (`bins`) y las **etiquetas** (`labels`).

```python
# --- pd.cut: convertir una variable numerica en categorias ordenadas ---
df["segmento_precio"] = pd.cut(
    df["precio"],
    bins=[0, 100000, 500000, 1500000, np.inf],          # los limites de cada intervalo
    labels=["Economico", "Medio", "Alto", "Premium"]     # una etiqueta menos que limites
)

print(df["segmento_precio"].value_counts().sort_index())

# qcut divide en grupos de igual CANTIDAD de registros (cuartiles), no de igual amplitud
df["cuartil_ingreso"] = pd.qcut(df["ingreso_neto"], q=4, labels=["Q1", "Q2", "Q3", "Q4"])
print("\nDistribucion por cuartiles (cada grupo tiene ~25% de los registros):")
print(df["cuartil_ingreso"].value_counts().sort_index())
```

<div style="background:#FDF3D8;border-left:6px solid #F2C14E;border-radius:6px;padding:14px 18px;margin:14px 0;">
  <div style="font-weight:700;color:#4A3708;font-size:14px;margin-bottom:4px;"><code>cut</code> vs <code>qcut</code>: no son intercambiables</div>
  <div style="color:#4A3708;font-size:14px;line-height:1.6;"><code>cut</code> corta por <b>amplitud</b>: los intervalos son del tamano que tu definas, y un grupo puede quedar con muy pocos registros. <code>qcut</code> corta por <b>cantidad</b>: fuerza a que cada grupo tenga aproximadamente el mismo numero de registros, y los intervalos resultan de amplitud desigual. Usa <code>cut</code> cuando los cortes tienen significado de negocio (menos de 100 mil = economico); usa <code>qcut</code> cuando quieres comparar el 25% mas alto con el 25% mas bajo.</div>
</div>


### `map()` y `apply()`: cuando necesitas logica propia

- **`.map()`** funciona sobre una **Series** y sirve para traducir valores uno a uno usando un diccionario. Es lo que usamos al inicio para deducir la region a partir de la ciudad.
- **`.apply()`** aplica una funcion a cada elemento (si se usa sobre una Series) o a cada fila/columna completa (si se usa sobre un DataFrame con `axis=1`).

```python
# --- map(): traducir valores con un diccionario ---
traduccion = {"Computo": "Computing", "Audio": "Audio", "Accesorios": "Accessories", "Telefonia": "Phones"}
df["categoria_en"] = df["categoria"].map(traduccion)
print(df[["categoria", "categoria_en"]].drop_duplicates())

# --- apply() sobre una Series: aplicar una funcion a cada valor ---
def clasificar_calificacion(nota):
    if pd.isna(nota):
        return "Sin calificar"
    elif nota >= 4:
        return "Satisfecho"
    elif nota == 3:
        return "Neutral"
    else:
        return "Insatisfecho"

df["satisfaccion"] = df["calificacion"].apply(clasificar_calificacion)
print("\n", df["satisfaccion"].value_counts())

# --- apply() con axis=1: la funcion recibe la FILA completa ---
df["alerta"] = df.apply(
    lambda fila: "REVISAR" if (fila["descuento"] >= 0.15 and fila["ingreso_neto"] > 3000000) else "OK",
    axis=1
)
print("\nVentas marcadas para revision:", (df["alerta"] == "REVISAR").sum())
```

Con 300 filas no lo notaras; con 3 millones puede tardar minutos frente a segundos. <b>Regla practica:</b> si lo que necesitas se puede expresar con operaciones aritmeticas, <code>np.where</code>, <code>map</code> o metodos <code>.str</code>, usa esos. Reserva <code>apply()</code> para logica que genuinamente no se puede vectorizar.</div>

Ordenar parece trivial, pero hay matices que importan: por que columna, en que direccion, que hacer con los empates y donde ubicar los nulos.

- **`sort_values(by=...)`** ordena por los valores de una o varias columnas.
- **`sort_index()`** ordena por el indice.
- **`nlargest(n, col)` / `nsmallest(n, col)`** son atajos para "los N mayores/menores": mas rapidos y mas legibles que ordenar todo y luego cortar.

```python
# --- Ordenar por una columna, de mayor a menor ---
top = df.sort_values(by="ingreso_neto", ascending=False)
display(top[["ciudad", "categoria", "unidades", "precio", "ingreso_neto"]].head())

# --- Ordenar por VARIAS columnas con direcciones distintas ---
# Primero por region (A-Z), y dentro de cada region por ingreso (mayor a menor)
multi = df.sort_values(by=["region", "ingreso_neto"], ascending=[True, False])
display(multi[["region", "ciudad", "ingreso_neto"]].head(8))

# --- Atajo para los N mayores ---
print("Las 3 ventas mas grandes (nlargest):")
display(df.nlargest(3, "ingreso_neto")[["ciudad", "categoria", "ingreso_neto"]])
```

`groupby` implementa un patron llamado **split-apply-combine** (dividir, aplicar, combinar), y entenderlo conceptualmente vale mas que memorizar la sintaxis:

1. **Split (dividir)**: pandas separa el DataFrame en grupos segun los valores de una columna. Si agrupas por `ciudad` y hay 6 ciudades, se forman 6 sub-tablas.
2. **Apply (aplicar)**: sobre cada sub-tabla se calcula una funcion de resumen (suma, promedio, conteo...).
3. **Combine (combinar)**: los resultados se unen en un solo objeto, con los valores de agrupacion como indice.

La pregunta que responde `groupby` siempre tiene esta forma: **"cual es el [resumen] de [variable] por cada [categoria]"**. Si puedes formular tu pregunta asi, ya sabes que codigo escribir.

```python
# --- groupby basico: una columna de agrupacion, una funcion de resumen ---
print("Ingreso neto TOTAL por ciudad:")
print(df.groupby("ciudad", observed=True)["ingreso_neto"].sum().sort_values(ascending=False))

print("\nIngreso PROMEDIO por categoria:")
print(df.groupby("categoria", observed=True)["ingreso_neto"].mean().round(0))

print("\nCANTIDAD de ventas por region:")
print(df.groupby("region", observed=True).size())
```

En pandas 2.x conviene ponerlo explicitamente porque el valor por defecto esta cambiando.</div>

### `agg()`: varios calculos a la vez

Rara vez quieres un solo estadistico. `agg()` permite pedir varios de golpe, y en su forma mas util (*named aggregation*) te deja **nombrar cada columna del resultado**, lo que produce tablas listas para presentar.

La sintaxis es: `nombre_resultado=("columna_origen", "funcion")`.

```python
# --- agg con nombres explicitos: la forma mas util y legible ---
resumen = df.groupby("ciudad", observed=True).agg(
    numero_ventas    = ("id_venta", "count"),
    unidades_totales = ("unidades", "sum"),
    ingreso_total    = ("ingreso_neto", "sum"),
    ticket_promedio  = ("ingreso_neto", "mean"),
    ticket_maximo    = ("ingreso_neto", "max"),
    calificacion_med = ("calificacion", "mean"),
).round(2).sort_values("ingreso_total", ascending=False)

display(resumen)
```

Ese resultado ya es un **informe de gestion**: en una sola instruccion respondiste cuantas ventas, cuantas unidades, cuanto ingreso, cual es el ticket promedio y como califican los clientes en cada ciudad. Piensa cuanto codigo requeriria esto con ciclos de Python puro.

### Agrupar por varias columnas

Pasando una lista a `groupby` obtienes un analisis cruzado. El resultado tiene un **indice jerarquico** (MultiIndex): dos niveles de agrupacion.

```python
# --- Agrupacion por DOS columnas ---
cruzado = df.groupby(["region", "categoria"], observed=True).agg(
    ventas  = ("id_venta", "count"),
    ingreso = ("ingreso_neto", "sum"),
).round(0)

display(cruzado.head(10))

# reset_index() convierte el indice jerarquico en columnas normales: mas facil de exportar y graficar
print("El mismo resultado en formato de tabla plana:")
display(cruzado.reset_index().head())
```

### `pivot_table()`: el equivalente a las tablas dinamicas de Excel

Cuando agrupas por dos variables, muchas veces prefieres verlas como **matriz**: una variable en las filas, otra en las columnas. Eso es exactamente una tabla dinamica, y `pivot_table` la construye.

Sus cuatro parametros clave: `index` (que va en las filas), `columns` (que va en las columnas), `values` (que se calcula) y `aggfunc` (como se calcula).

```python
# --- pivot_table: tabla dinamica ---
pivote = pd.pivot_table(
    df,
    index="ciudad",          # filas
    columns="categoria",     # columnas
    values="ingreso_neto",   # valores a resumir
    aggfunc="sum",           # como resumirlos
    fill_value=0,            # que poner donde no hay datos
    margins=True,            # agrega fila y columna de totales
    margins_name="TOTAL",
    observed=True,
)

display(pivote.round(0))
```

### `transform()`: agregar sin perder el detalle

A veces necesitas comparar cada fila con el resumen de su grupo (por ejemplo, "que porcentaje del total de su ciudad representa esta venta"). `groupby().transform()` calcula el agregado pero **devuelve un resultado del mismo tamano que el original**, listo para asignarlo como columna.

```python
# --- transform: el agregado del grupo, alineado con cada fila ---
df["ingreso_ciudad"] = df.groupby("ciudad", observed=True)["ingreso_neto"].transform("sum")
df["participacion_pct"] = (df["ingreso_neto"] / df["ingreso_ciudad"] * 100).round(2)

display(df[["ciudad", "ingreso_neto", "ingreso_ciudad", "participacion_pct"]].head())

print("Nota: 'ingreso_ciudad' repite el total de la ciudad en cada fila de esa ciudad.")
print("Eso es justamente lo que permite calcular la participacion individual.")
```

En cualquier organizacion la informacion esta repartida: las ventas en un sistema, los clientes en otro, las metas en una hoja de calculo. Combinar tablas es una tarea cotidiana y hay dos operaciones distintas que conviene no confundir:

- **`concat()`**: **apila**. Une tablas que tienen la misma estructura, agregando filas (o columnas) una tras otra. Piensa en juntar las ventas de enero, febrero y marzo.
- **`merge()`**: **cruza**. Une tablas que comparten una **llave** (un identificador comun), trayendo columnas de una tabla a la otra. Es el equivalente al `JOIN` de SQL y al `BUSCARV`/`VLOOKUP` de Excel.

### `concat()`: apilar

```python
# --- concat: apilar tablas con la misma estructura ---
enero   = pd.DataFrame({"mes": ["ene", "ene"], "ciudad": ["Bogota", "Cali"], "ventas": [100, 80]})
febrero = pd.DataFrame({"mes": ["feb", "feb"], "ciudad": ["Bogota", "Cali"], "ventas": [120, 95]})

trimestre = pd.concat([enero, febrero], ignore_index=True)
display(trimestre)

print("ignore_index=True renumera el indice; sin el quedaria 0,1,0,1 y causaria confusion.")
```

### `merge()`: cruzar por una llave

El parametro decisivo de `merge` es **`how`**, que define **que filas sobreviven al cruce**. Esta es la tabla que debes tener clara:


<div style="overflow-x:auto;">
<table style="width:100%;border-collapse:collapse;font-size:14px;">
  <tr style="background:#0B4F6C;color:#FFFFFF;"><th style="padding:10px;text-align:left;"><code>how</code></th><th style="padding:10px;text-align:left;">Que filas conserva</th><th style="padding:10px;text-align:left;">Cuando usarlo</th></tr>
  <tr style="background:#F4F6F8;color:#242B33;"><td style="padding:9px;vertical-align:top;"><code>'inner'</code> (defecto)</td><td style="padding:9px;vertical-align:top;">Solo las llaves que estan en <b>ambas</b> tablas</td><td style="padding:9px;vertical-align:top;">Cuando solo te interesan los registros completos</td></tr><tr style="color:#242B33;"><td style="padding:9px;vertical-align:top;"><code>'left'</code></td><td style="padding:9px;vertical-align:top;"><b>Todas</b> las de la tabla izquierda; rellena con <code>NaN</code> lo que no encuentre</td><td style="padding:9px;vertical-align:top;">El mas usado: enriquecer tu tabla principal sin perder filas</td></tr><tr style="background:#F4F6F8;color:#242B33;"><td style="padding:9px;vertical-align:top;"><code>'right'</code></td><td style="padding:9px;vertical-align:top;">Todas las de la derecha</td><td style="padding:9px;vertical-align:top;">Poco usado; equivale a un <code>left</code> invirtiendo el orden</td></tr><tr style="color:#242B33;"><td style="padding:9px;vertical-align:top;"><code>'outer'</code></td><td style="padding:9px;vertical-align:top;">Todas las de ambas, con <code>NaN</code> donde falte</td><td style="padding:9px;vertical-align:top;">Cuando necesitas ver que hay en cada lado y que no cruza</td></tr>
</table>
</div>

```python
# --- Tabla auxiliar: metas mensuales por ciudad ---
metas = pd.DataFrame({
    "ciudad": ["Bogota", "Medellin", "Cali", "Barranquilla", "Cartagena", "Pereira"],
    "meta_anual": [800_000_000, 500_000_000, 450_000_000, 300_000_000, 350_000_000, 200_000_000],
    "gerente": ["L. Ramos", "M. Pena", "J. Duque", "S. Vega", "P. Rojas", "T. Ortiz"],
})
display(metas)

# Resumen real de ingresos por ciudad
real = df.groupby("ciudad", observed=True)["ingreso_neto"].sum().reset_index()
real.columns = ["ciudad", "ingreso_real"]

# --- merge LEFT: conservamos todas las ciudades con ventas ---
comparacion = real.merge(metas, on="ciudad", how="left")
comparacion["cumplimiento_pct"] = (comparacion["ingreso_real"] / comparacion["meta_anual"] * 100).round(1)

display(comparacion.sort_values("cumplimiento_pct", ascending=False))
```

Observa un detalle revelador: **Bucaramanga tiene ventas pero no aparece en la tabla de metas, y Pereira tiene meta pero no ventas**. Con `how="left"` desde `real`, Bucaramanga aparece con meta nula y Pereira desaparece. Cambiar el `how` cambia por completo la historia que cuentan los datos. Veamoslo explicitamente, porque este es el punto donde mas errores silenciosos se cometen.

```python
# --- Comparar el efecto de cada tipo de union ---
for tipo in ["inner", "left", "right", "outer"]:
    r = real.merge(metas, on="ciudad", how=tipo)
    print(f"how='{tipo:6s}' -> {len(r):2d} filas | ciudades: {sorted(r['ciudad'].tolist())}")

print("\nLeccion: elegir mal el 'how' hace desaparecer registros SIN AVISAR.")
print("Verifica SIEMPRE el numero de filas antes y despues de un merge.")
```

  <div style="font-weight:700;color:#7A1A14;font-size:14px;margin-bottom:4px;">Verificacion obligatoria despues de un merge</div>
  <div style="color:#7A1A14;font-size:14px;line-height:1.6;">Compara <code>len(df)</code> antes y despues. Si el resultado tiene <b>menos</b> filas, perdiste registros que no cruzaron. Si tiene <b>mas</b>, la llave estaba duplicada en la tabla derecha y se multiplicaron las filas: un error muy comun que infla todas las sumas posteriores. El parametro <code>validate='m:1'</code> hace que pandas lance un error si la relacion no es la que esperabas.</div>
</div>

```python
# --- validate: que pandas verifique la relacion por ti ---
try:
    # 'm:1' significa: muchas filas de la izquierda pueden cruzar con UNA sola de la derecha
    ok = real.merge(metas, on="ciudad", how="left", validate="m:1")
    print("Validacion superada: la llave 'ciudad' es unica en la tabla de metas.")
except Exception as e:
    print("La validacion fallo:", e)

# Si las columnas llave tienen NOMBRES distintos en cada tabla:
# df1.merge(df2, left_on="codigo_ciudad", right_on="id_ciudad", how="left")
```

Mientras una fecha sea texto, no puedes hacer nada util con ella: no puedes ordenar cronologicamente, ni calcular diferencias, ni agrupar por mes. El primer paso es **convertirla al tipo `datetime`** con `pd.to_datetime()`.

Una vez convertida, el accesor **`.dt`** abre acceso a todos sus componentes, igual que `.str` lo hace con el texto.

```python
# --- Verificar el tipo de la columna de fecha ---
print("dtype de 'fecha':", df["fecha"].dtype)

# Si fuera texto, se convertiria asi:
# df["fecha"] = pd.to_datetime(df["fecha"], format="%d/%m/%Y", errors="coerce")

# --- El accesor .dt extrae componentes de la fecha ---
df["ano"]       = df["fecha"].dt.year
df["mes"]       = df["fecha"].dt.month
df["nombre_mes"] = df["fecha"].dt.month_name()
df["dia_semana"] = df["fecha"].dt.day_name()
df["trimestre"] = df["fecha"].dt.quarter
df["semana"]    = df["fecha"].dt.isocalendar().week

display(df[["fecha", "ano", "mes", "nombre_mes", "dia_semana", "trimestre", "semana"]].head())
```

```python
# --- Analisis temporal: ingreso por mes ---
por_mes = df.groupby("mes")["ingreso_neto"].sum().round(0)
print("Ingreso por mes:")
print(por_mes)

print("\nMejor mes:", por_mes.idxmax(), "con $", f"{por_mes.max():,.0f}")

# --- Diferencias entre fechas ---
print("\nRango del dataset:", df["fecha"].min().date(), "a", df["fecha"].max().date())
print("Dias cubiertos:", (df["fecha"].max() - df["fecha"].min()).days)

# --- resample: agregar por periodo (requiere la fecha como indice) ---
serie_mensual = df.set_index("fecha")["ingreso_neto"].resample("ME").sum()
print("\nSerie mensual con resample('ME') = fin de mes:")
print(serie_mensual.head())
```

### Error 1: olvidar que pandas devuelve copias

Ya lo mencionamos, pero es tan frecuente que merece su propia seccion. Casi todos los metodos de pandas **devuelven un objeto nuevo** y dejan el original intacto.

```python
# --- Demostracion del error mas comun ---
prueba = pd.DataFrame({"a": [1, 2, 3], "b": [4, 5, 6]})

prueba.drop(columns="b")     # <- INCORRECTO: el resultado se pierde
print("Columnas despues del drop sin asignar:", list(prueba.columns), "<- la 'b' sigue ahi")

prueba = prueba.drop(columns="b")   # <- CORRECTO: reasignamos
print("Columnas despues de reasignar     :", list(prueba.columns))
```

Reasignar es mas claro y mas seguro.</div>

### Error 2: el `SettingWithCopyWarning`

Este aviso aparece cuando modificas un DataFrame que pandas no sabe con certeza si es una copia independiente o una vista del original. El resultado puede ser que tu cambio **no se aplique donde esperas**, silenciosamente.

La causa es la **indexacion encadenada**: dos operaciones de seleccion seguidas.

```python
# --- Forma INCORRECTA: indexacion encadenada ---
sub = df[df["ciudad"] == "Bogota"]      # esto puede ser una vista del original
# sub["nueva"] = 1                      # <- generaria SettingWithCopyWarning

# --- Forma CORRECTA 1: pedir una copia explicita ---
sub = df[df["ciudad"] == "Bogota"].copy()
sub["nueva"] = 1
print("Con .copy() no hay advertencia. Filas de Bogota:", len(sub))

# --- Forma CORRECTA 2: modificar el original con .loc en UNA sola operacion ---
df["es_capital"] = False                                  # se crea la columna con un valor por defecto
df.loc[df["ciudad"] == "Bogota", "es_capital"] = True      # se actualizan solo las filas que cumplen
print("Registros marcados como capital:", df["es_capital"].sum())
```

### Error 3: confundir `NaN` con cero, y comparar nulos con `==`

`NaN` tiene una propiedad matematica peculiar: **no es igual a si mismo**. Por eso `df["col"] == np.nan` nunca da `True` y hay que usar `.isna()`.

Ademas, las funciones de agregacion **ignoran los nulos por defecto**, lo que puede producir resultados que no esperas si no lo tienes presente.

```python
# --- NaN no es igual a NaN ---
print("np.nan == np.nan  ->", np.nan == np.nan, " <- por eso NUNCA compares con ==")
print("pd.isna(np.nan)   ->", pd.isna(np.nan), " <- esta es la forma correcta")

# --- Efecto de los nulos en las agregaciones ---
s = pd.Series([10, 20, np.nan, 40])
print("\nSerie:", list(s))
print("sum()          :", s.sum(), "  <- ignora el nulo")
print("mean()         :", s.mean(), " <- divide entre 3, NO entre 4")
print("count()        :", s.count(), "     <- cuenta solo los NO nulos")
print("len()          :", len(s), "     <- cuenta TODAS las posiciones")
print("mean(skipna=False):", s.mean(skipna=False), "<- si quieres que el nulo contamine el resultado")
```

```python
# ============================================================
#  DATASET DE LOS EJERCICIOS - EJECUTA ESTA CELDA PRIMERO
#  Prestamos de la biblioteca universitaria, semestre 2024-1
# ============================================================
import pandas as pd
import numpy as np

np.random.seed(2024)
m = 420

facultades = ["Ingenieria", "Administracion", "Ciencias", "Artes", "Derecho"]
generos    = ["Novela", "Tecnico", "Referencia", "Revista", "Ensayo", "Poesia"]
sedes      = ["Sede Norte", "Sede Centro", "Sede Sur"]

biblioteca = pd.DataFrame({
    "id_prestamo":  range(5001, 5001 + m),
    "codigo_libro": np.random.choice([f"L-{i:04d}" for i in range(1, 121)], m),
    "genero":       np.random.choice(generos, m, p=[.22, .28, .12, .15, .13, .10]),
    "sede":         np.random.choice(sedes, m, p=[.45, .35, .20]),
    "facultad":     np.random.choice(facultades, m),
    "semestre_estudiante": np.random.randint(1, 11, m),
    "fecha_prestamo":  pd.to_datetime("2024-02-01") + pd.to_timedelta(np.random.randint(0, 120, m), unit="D"),
    "dias_prestado":   np.random.choice([3, 5, 7, 10, 14, 21, 30], m, p=[.10, .18, .27, .18, .15, .08, .04]),
    "renovaciones":    np.random.choice([0, 1, 2, 3], m, p=[.55, .27, .13, .05]),
    "multa":           np.round(np.random.choice([0, 0, 0, 0, 1500, 3200, 7800, 15000], m), 0),
    "satisfaccion":    np.random.choice([1, 2, 3, 4, 5, np.nan], m, p=[.04, .08, .20, .38, .25, .05]),
})

# Defectos introducidos a proposito: son parte del ejercicio
biblioteca.loc[np.random.choice(biblioteca.index, 18, replace=False), "facultad"] = np.nan
biblioteca.loc[np.random.choice(biblioteca.index, 9, replace=False), "sede"] = "  sede norte  "
biblioteca["dias_prestado"] = biblioteca["dias_prestado"].astype(str)
biblioteca = pd.concat([biblioteca, biblioteca.sample(11, random_state=5)], ignore_index=True)

# Tabla auxiliar para los ejercicios de cruce
catalogo = pd.DataFrame({
    "codigo_libro": [f"L-{i:04d}" for i in range(1, 111)],   # ojo: no estan los 120
    "titulo":       [f"Titulo {i}" for i in range(1, 111)],
    "anio_edicion": np.random.randint(1998, 2024, 110),
    "ejemplares":   np.random.randint(1, 9, 110),
})

print("Dataset de ejercicios cargado.")
print("biblioteca:", biblioteca.shape, "| catalogo:", catalogo.shape)
biblioteca.head()
```
